```python
# ============================================================
# ASSIGNMENT: ENSEMBLE METHODS
# Dataset: Titanic
# Models:
# 1. Logistic Regression (Earlier Single Model)
# 2. Random Forest Classifier
# 3. XGBoost Classifier
# ============================================================


# ============================================================
# 1. INSTALL XGBOOST
# ============================================================

!pip install -q xgboost


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

print("=" * 70)
print("ENSEMBLE METHODS - TITANIC CLASSIFICATION")
print("=" * 70)


# ============================================================
# 3. LOAD TITANIC DATASET
# ============================================================

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())


# ============================================================
# 4. BASIC EDA
# ============================================================

print("\n" + "=" * 70)
print("BASIC EDA")
print("=" * 70)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe(include="all").T)


# ============================================================
# 5. CREATE ENGINEERED FEATURES
# ============================================================

print("\n" + "=" * 70)
print("FEATURE ENGINEERING")
print("=" * 70)

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

df["IsAlone"] = (
    df["FamilySize"] == 1
).astype(int)

print("Created features:")
print("1. FamilySize = SibSp + Parch + 1")
print("2. IsAlone = 1 if FamilySize == 1, otherwise 0")

display(
    df[
        ["SibSp", "Parch", "FamilySize", "IsAlone", "Survived"]
    ].head(10)
)


# ============================================================
# 6. VISUALIZE IMPORTANT RELATIONSHIPS
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()


plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived"
)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()


plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="FamilySize",
    hue="Survived"
)

plt.title("Survival by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Number of Passengers")
plt.show()


# ============================================================
# 7. SELECT FEATURES
# ============================================================

features = [
    "Pclass",
    "Sex",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "Embarked",
    "FamilySize",
    "IsAlone"
]

X = df[features]
y = df["Survived"]


# ============================================================
# 8. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])


# ============================================================
# 9. DEFINE NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

numerical_features = [
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

categorical_features = [
    "Sex",
    "Embarked"
]


# ============================================================
# 10. PREPROCESSING
# ============================================================

numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)


categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "onehot",
            OneHotEncoder(handle_unknown="ignore")
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numerical_transformer,
            numerical_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ============================================================
# 11. MODEL 1 - LOGISTIC REGRESSION
# ============================================================

print("\n" + "=" * 70)
print("MODEL 1 - LOGISTIC REGRESSION")
print("=" * 70)

logistic_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)

logistic_pipeline.fit(
    X_train,
    y_train
)

logistic_pred = logistic_pipeline.predict(
    X_test
)


# Logistic Regression Metrics
logistic_accuracy = accuracy_score(
    y_test,
    logistic_pred
)

logistic_precision = precision_score(
    y_test,
    logistic_pred
)

logistic_recall = recall_score(
    y_test,
    logistic_pred
)

logistic_f1 = f1_score(
    y_test,
    logistic_pred
)

print("\nLogistic Regression Performance:")
print(f"Accuracy : {logistic_accuracy:.4f}")
print(f"Precision: {logistic_precision:.4f}")
print(f"Recall   : {logistic_recall:.4f}")
print(f"F1 Score : {logistic_f1:.4f}")


# ============================================================
# 12. MODEL 2 - RANDOM FOREST
# ============================================================

print("\n" + "=" * 70)
print("MODEL 2 - RANDOM FOREST")
print("=" * 70)


random_forest_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestClassifier(
                n_estimators=200,
                max_depth=None,
                min_samples_split=2,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


random_forest_pipeline.fit(
    X_train,
    y_train
)


rf_pred = random_forest_pipeline.predict(
    X_test
)


# Random Forest Metrics
rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_precision = precision_score(
    y_test,
    rf_pred
)

rf_recall = recall_score(
    y_test,
    rf_pred
)

rf_f1 = f1_score(
    y_test,
    rf_pred
)

print("\nRandom Forest Performance:")
print(f"Accuracy : {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall   : {rf_recall:.4f}")
print(f"F1 Score : {rf_f1:.4f}")


# ============================================================
# 13. MODEL 3 - XGBOOST
# ============================================================

print("\n" + "=" * 70)
print("MODEL 3 - XGBOOST")
print("=" * 70)


xgb_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            XGBClassifier(
                n_estimators=200,
                max_depth=4,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                objective="binary:logistic",
                eval_metric="logloss",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


xgb_pipeline.fit(
    X_train,
    y_train
)


xgb_pred = xgb_pipeline.predict(
    X_test
)


# XGBoost Metrics
xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)

xgb_precision = precision_score(
    y_test,
    xgb_pred
)

xgb_recall = recall_score(
    y_test,
    xgb_pred
)

xgb_f1 = f1_score(
    y_test,
    xgb_pred
)


print("\nXGBoost Performance:")
print(f"Accuracy : {xgb_accuracy:.4f}")
print(f"Precision: {xgb_precision:.4f}")
print(f"Recall   : {xgb_recall:.4f}")
print(f"F1 Score : {xgb_f1:.4f}")


# ============================================================
# 14. COMPLETE MODEL COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)


comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],
    "Accuracy": [
        logistic_accuracy,
        rf_accuracy,
        xgb_accuracy
    ],
    "Precision": [
        logistic_precision,
        rf_precision,
        xgb_precision
    ],
    "Recall": [
        logistic_recall,
        rf_recall,
        xgb_recall
    ],
    "F1 Score": [
        logistic_f1,
        rf_f1,
        xgb_f1
    ]
})


display(
    comparison.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)


# ============================================================
# 15. FIND BEST MODEL
# ============================================================

best_model_index = comparison["F1 Score"].idxmax()

best_model = comparison.loc[
    best_model_index,
    "Model"
]

best_f1 = comparison.loc[
    best_model_index,
    "F1 Score"
]

print(
    f"\nBest model based on F1 Score: "
    f"{best_model}"
)

print(
    f"Best F1 Score: {best_f1:.4f}"
)


# ============================================================
# 16. CLASSIFICATION REPORTS
# ============================================================

print("\n" + "=" * 70)
print("LOGISTIC REGRESSION CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        logistic_pred,
        target_names=[
            "Not Survived",
            "Survived"
        ]
    )
)


print("\n" + "=" * 70)
print("RANDOM FOREST CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        rf_pred,
        target_names=[
            "Not Survived",
            "Survived"
        ]
    )
)


print("\n" + "=" * 70)
print("XGBOOST CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        xgb_pred,
        target_names=[
            "Not Survived",
            "Survived"
        ]
    )
)


# ============================================================
# 17. CONFUSION MATRICES
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(18, 5)
)


# Logistic Regression
sns.heatmap(
    confusion_matrix(y_test, logistic_pred),
    annot=True,
    fmt="d",
    ax=axes[0]
)

axes[0].set_title(
    "Logistic Regression"
)

axes[0].set_xlabel(
    "Predicted"
)

axes[0].set_ylabel(
    "Actual"
)


# Random Forest
sns.heatmap(
    confusion_matrix(y_test, rf_pred),
    annot=True,
    fmt="d",
    ax=axes[1]
)

axes[1].set_title(
    "Random Forest"
)

axes[1].set_xlabel(
    "Predicted"
)

axes[1].set_ylabel(
    "Actual"
)


# XGBoost
sns.heatmap(
    confusion_matrix(y_test, xgb_pred),
    annot=True,
    fmt="d",
    ax=axes[2]
)

axes[2].set_title(
    "XGBoost"
)

axes[2].set_xlabel(
    "Predicted"
)

axes[2].set_ylabel(
    "Actual"
)


plt.tight_layout()
plt.show()


# ============================================================
# 18. FEATURE IMPORTANCE - RANDOM FOREST
# ============================================================

print("\n" + "=" * 70)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 70)


# Get transformed feature names
feature_names = (
    random_forest_pipeline
    .named_steps["preprocessor"]
    .get_feature_names_out()
)


# Get Random Forest model
rf_model = (
    random_forest_pipeline
    .named_steps["model"]
)


# Feature importance
rf_importance = rf_model.feature_importances_


rf_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_importance
})


rf_importance_df = (
    rf_importance_df
    .sort_values(
        "Importance",
        ascending=False
    )
)


print("\nTop Random Forest Features:")
display(
    rf_importance_df.head(15)
)


# Plot
plt.figure(figsize=(10, 7))

sns.barplot(
    data=rf_importance_df.head(15),
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 15 Random Forest Feature Importances"
)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()
plt.show()


# ============================================================
# 19. FEATURE IMPORTANCE - XGBOOST
# ============================================================

print("\n" + "=" * 70)
print("XGBOOST FEATURE IMPORTANCE")
print("=" * 70)


# Get XGBoost model
xgb_model = (
    xgb_pipeline
    .named_steps["model"]
)


# Feature importance
xgb_importance = xgb_model.feature_importances_


xgb_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": xgb_importance
})


xgb_importance_df = (
    xgb_importance_df
    .sort_values(
        "Importance",
        ascending=False
    )
)


print("\nTop XGBoost Features:")
display(
    xgb_importance_df.head(15)
)


# Plot
plt.figure(figsize=(10, 7))

sns.barplot(
    data=xgb_importance_df.head(15),
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 15 XGBoost Feature Importances"
)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()
plt.show()


# ============================================================
# 20. COMPARE FEATURE IMPORTANCES
# ============================================================

print("\n" + "=" * 70)
print("FEATURE IMPORTANCE COMPARISON")
print("=" * 70)


top_rf = rf_importance_df.head(10)[
    ["Feature", "Importance"]
].copy()

top_rf["Model"] = "Random Forest"


top_xgb = xgb_importance_df.head(10)[
    ["Feature", "Importance"]
].copy()

top_xgb["Model"] = "XGBoost"


feature_comparison = pd.concat(
    [top_rf, top_xgb],
    ignore_index=True
)


display(
    feature_comparison.round(4)
)


# ============================================================
# 21. COMBINED FEATURE IMPORTANCE PLOT
# ============================================================

# Select top features from both models
top_features = list(
    set(
        rf_importance_df.head(10)["Feature"]
    ).union(
        set(
            xgb_importance_df.head(10)["Feature"]
        )
    )
)


rf_plot_data = (
    rf_importance_df[
        rf_importance_df["Feature"].isin(
            top_features
        )
    ]
    .set_index("Feature")
    .reindex(top_features)
    .fillna(0)
)


xgb_plot_data = (
    xgb_importance_df[
        xgb_importance_df["Feature"].isin(
            top_features
        )
    ]
    .set_index("Feature")
    .reindex(top_features)
    .fillna(0)
)


importance_comparison = pd.DataFrame({
    "Random Forest": rf_plot_data["Importance"],
    "XGBoost": xgb_plot_data["Importance"]
})


importance_comparison = (
    importance_comparison
    .sort_values(
        "XGBoost",
        ascending=False
    )
)


plt.figure(figsize=(12, 8))

importance_comparison.plot(
    kind="barh",
    figsize=(12, 8)
)

plt.title(
    "Random Forest vs XGBoost Feature Importance"
)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()


# ============================================================
# 22. SAVE MODELS
# ============================================================

print("\n" + "=" * 70)
print("SAVING MODELS")
print("=" * 70)


joblib.dump(
    logistic_pipeline,
    "titanic_logistic_regression.pkl"
)

joblib.dump(
    random_forest_pipeline,
    "titanic_random_forest.pkl"
)

joblib.dump(
    xgb_pipeline,
    "titanic_xgboost.pkl"
)


print("✓ Logistic Regression saved")
print("✓ Random Forest saved")
print("✓ XGBoost saved")


# ============================================================
# 23. VERIFY SAVED MODELS
# ============================================================

loaded_rf = joblib.load(
    "titanic_random_forest.pkl"
)

loaded_xgb = joblib.load(
    "titanic_xgboost.pkl"
)


loaded_rf_pred = loaded_rf.predict(X_test)
loaded_xgb_pred = loaded_xgb.predict(X_test)


print("\nRandom Forest loaded successfully:",
      np.array_equal(rf_pred, loaded_rf_pred))

print("XGBoost loaded successfully:",
      np.array_equal(xgb_pred, loaded_xgb_pred))


# ============================================================
# 24. GITHUB-READY COMPARISON TABLE
# ============================================================

github_table = comparison.copy()

github_table["Accuracy"] = (
    github_table["Accuracy"]
    .round(4)
)

github_table["Precision"] = (
    github_table["Precision"]
    .round(4)
)

github_table["Recall"] = (
    github_table["Recall"]
    .round(4)
)

github_table["F1 Score"] = (
    github_table["F1 Score"]
    .round(4)
)


print("\n" + "=" * 70)
print("GITHUB COMPARISON TABLE")
print("=" * 70)

display(github_table)


# Print Markdown table
print("\nMarkdown version for README.md:\n")

print(
    github_table.to_markdown(
        index=False
    )
)


# ============================================================
# 25. FINAL CONCLUSION
# ============================================================

print("\n" + "=" * 70)
print("FINAL CONCLUSION")
print("=" * 70)

print("""
Random Forest and XGBoost are ensemble learning methods that combine
multiple decision trees to produce stronger predictions than a single
tree. Random Forest builds many trees independently using random samples
and random subsets of features, and then combines their predictions
through voting for classification. XGBoost builds trees sequentially,
where each new tree focuses on correcting the errors made by previous
trees using gradient boosting. As a result, Random Forest generally
reduces variance through averaging, while XGBoost focuses on reducing
errors iteratively through boosting.
""")

print(
    f"Based on F1 Score, the best performing model was: {best_model}"
)

print(
    f"Best F1 Score: {best_f1:.4f}"
)

print("""
The models were evaluated using Accuracy, Precision, Recall, and F1 Score.
Feature importance plots were also generated for Random Forest and XGBoost
to compare which features were considered most influential.

All trained models were saved using joblib and can be reused without
retraining.
""")


# ============================================================
# 26. FINAL FILE LIST
# ============================================================

print("\n" + "=" * 70)
print("FILES CREATED")
print("=" * 70)

print("""
1. titanic_logistic_regression.pkl
2. titanic_random_forest.pkl
3. titanic_xgboost.pkl

These files contain the complete preprocessing + trained model pipelines.
""")

print("=" * 70)
print("ASSIGNMENT COMPLETED SUCCESSFULLY!")
print("=" * 70)
```

Random Forest builds many decision trees independently and combines their predictions through voting, which helps reduce overfitting and variance. XGBoost also combines decision trees, but it builds them sequentially, with each new tree trying to correct the errors made by the previous trees. Random Forest mainly uses bagging and averaging, while XGBoost uses boosting and gradient optimization. This allows XGBoost to often achieve very strong predictive performance, although it generally requires more careful tuning.
